# Multi-Disease Clinical Risk Prediction: Exploratory Data Analysis (EDA)

This notebook conducts systematic exploratory data analysis across the three benchmark clinical datasets:
1. **Heart Disease Dataset** ($N = 10,000$, 20 clinical biomarkers)
2. **Diabetes Dataset** ($N = 768$, Pima Indians UCI benchmark)
3. **Breast Cancer Dataset** ($N = 569$, Wisconsin Diagnostic UCI benchmark)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams.update({"figure.autolayout": True})
print("Libraries loaded successfully.")
```

## 1. Data Ingestion & Shape Inspection

```python
df_heart = pd.read_csv("../data/raw/heart_disease.csv")
df_diab = pd.read_csv("../data/raw/diabetes.csv")
df_cancer = pd.read_csv("../data/raw/breast_cancer.csv")

print(f"Heart Disease shape: {df_heart.shape}")
print(f"Diabetes shape: {df_diab.shape}")
print(f"Breast Cancer shape: {df_cancer.shape}")
```

## 2. Target Class Distribution Analysis

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Heart disease balance
sns.countplot(x="Heart Disease Status", data=df_heart, ax=axes[0], palette=["#22c55e", "#ef4444"])
axes[0].set_title("Heart Disease Class Balance")

# Diabetes balance
sns.countplot(x="Outcome", data=df_diab, ax=axes[1], palette=["#22c55e", "#ef4444"])
axes[1].set_title("Diabetes Class Balance")

# Breast Cancer balance
sns.countplot(x="target", data=df_cancer, ax=axes[2], palette=["#ef4444", "#22c55e"])
axes[2].set_title("Breast Cancer Class Balance")
plt.show()
```

## 3. Detecting Physiological Zero-Value Anomalies in Diabetes

```python
zero_cols = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]
zero_counts = {col: (df_diab[col] == 0).sum() for col in zero_cols}
zero_series = pd.Series(zero_counts)

plt.figure(figsize=(8, 4))
bars = plt.bar(zero_series.index, zero_series.values, color="#f97316", edgecolor="#7c2d12")
plt.title("Invalid Zero Biomarkers in Diabetes Dataset")
plt.ylabel("Count of Zeros")
for bar in bars:
    h = bar.get_height()
    plt.annotate(f"{h} ({h/len(df_diab):.1%})", (bar.get_x() + bar.get_width() / 2, h + 8), ha="center")
plt.show()
```

## 4. Correlation Analysis

```python
plt.figure(figsize=(9, 7))
sns.heatmap(df_diab.corr(), annot=True, fmt=".2f", cmap="coolwarm", square=True)
plt.title("Diabetes Feature Correlation Heatmap")
plt.show()
```
